## Residential/commercial land use and crime
The study of crime and urban land use has given rise to different takes regarding the impacts of residential and commercial activity on neighbourhood crime rates. Theorists like Jane Jacobs are optimistic about the advantages of mixed land use, conferring the benefits of dispersing street activity, increasing "eyes on the street"-style social monitoring and control, and bolstering neighbourhood trust (Jacobs, 1961), all factors that can reduce the risk of crime. On the other hand, other approaches (e.g. Taylor, 1988) have associated the mixing of the residential and nonresidential with a heightened sense of anonymity and decreased territoriality and collective responsibility, thus increasing the propensity for crime. The latter position has contributed to general assumptions that residential land use exerts protective effects on crime, while commercial land use exacerbates it (Wo & Kim, 2020).

To what extent, however, is this assumption a simplistic take on the relationship between crime and land use? Investigating the relationship between land use and crime in the context of Columbus, Ohio, Browning et al. (2010) found it indeed nonlinear: while violent crime increased with commercial/residential density at low levels, it in fact decreased past when mixed land use increased beyond a threshold (except robbery). Similarly, Wo and Kim's (2020) study of residential versus nonresidential land use in San Francisco found a positive association between crime rates and commercial mixed-use areas, but found also that nonresidential land use exhibited non-significant and inconclusive associations with increased crime. In Zahnow's (2018) study of Brisbane, increased rates of crime was found to be tied not to mixed land use _within_ the study area, but in nearby areas, likely due to people being attracted to surrounding mixed-use areas with increased opportunities for interaction. In other words, nonresidential land use cannot always be associated with higher crime in the area (Stucky & Ottensmann, 2009), demonstrating the limitations of a binary understanding of residential/nonresidential land use and crime.

### 1. Violent crime data

```python
# reading in data from github repository
crimeurl = "https://raw.githubusercontent.com/jyjylow/casa0006-assessment/refs/heads/main/data/MPS%20LSOA%20Level%20Crime%20(most%20recent%2024%20months)%20(1).csv"
response = requests.get(crimeurl)
if response.status_code == 200:
    data = StringIO(response.text)
    crimedf = pd.read_csv(data)
    print("LSOA-level crime data loaded successfully.")
else:
    print(f"Failed to fetch the file: {response.status_code}")

# peeking at crime data
crimedf.head()
```

```python
# understanding the categories of crime reports
print(f"Counts of reports by major crime category:")
print(crimedf["Major Category"].value_counts().reset_index())
```

Clearly, "VIOLENCE AGAINST THE PERSON" reports are relevant to my study. Other categories might also be pertinent, so I'm taking a look at the subcategories so I don't miss out on important reports. 

```python
# displaying main and subcategories of crime reports in a table
crime_categories = crimedf[["Major Category", "Minor Category"]].drop_duplicates().sort_values(["Major Category", "Minor Category"]).reset_index(drop=True)
crime_categories
```

Based on this, I identified that:
- "PUBLIC ORDER OFFENCES: VIOLENT DISORDER"
- "VEHICLE OFFENCES: AGGRAVATED VEHICLE TAKING"

are also highly likely to be relevant to my study as they entail a degree of person-to-person physical violence. These will thus be included in analysis.

```python
# filtering crimedf to relevant rows/cols

# 1. filtering dataset to crimes of "VIOLENCE AGAINST THE PERSON" and the subcategories listed above
crimedf = crimedf[
    (crimedf["Major Category"] == "VIOLENCE AGAINST THE PERSON") |
    ((crimedf["Major Category"] == "PUBLIC ORDER OFFENCES") & (crimedf["Minor Category"] == "VIOLENT DISORDER")) |
    ((crimedf["Major Category"] == "VEHICLE OFFENCES") & (crimedf["Minor Category"] == "AGGRAVATED VEHICLE TAKING"))]
```

### 2a. London geometry

```python
# reading in London LSOA boundaries
lsoa_url = "https://github.com/jyjylow/casa0006-assessment/raw/main/data/statistical-gis-boundaries-london.zip"
lsoa_response = requests.get(lsoa_url)
with open("statistical-gis-boundaries-london.zip", "wb") as f:
    f.write(lsoa_response.content)

# extracting contents of zipped folder
with zipfile.ZipFile("statistical-gis-boundaries-london.zip", "r") as zip_ref:
    zip_ref.extractall("data/lsoa_shapefiles")
```

```python
# summing YYYYMM columns so that we have the aggregate sum of reports for past 24 months

# 1. identifying YYYYMM columns
monthly_cols = [col for col in crimedf.columns if col.isdigit()]

# 2. summing up YYYYMM columns
crimedf.loc[:, 'total_reports_past24'] = crimedf[monthly_cols].sum(axis=1)

# 3. filtering for relevant columns
crime_past24 = crimedf[["LSOA Code", "LSOA Name", "Major Category", "Minor Category", "total_reports_past24"]]
crime_past24.head()
```

```python
# summing YYYYMM columns so that we have yearly counts of crime reports for each ward

# 1. identifying YYYYMM columns
date_columns = [col for col in crimedf.columns if col.isdigit()]

# 2. summing columns that start with the same year
yearly_counts = pd.DataFrame()
for year in sorted(set(col[:4] for col in date_columns)):
    year_cols = [col for col in date_columns if col.startswith(year)]
   # Sum only across wards (rows), keeping crime categories separate
    yearly_counts[year] = crimedf.groupby(['LookUp_BoroughName', 'WardName', 'WardCode'])[year_cols].sum().sum(axis=1)

# 3. resetting index
yearly_counts = yearly_counts.reset_index()
print("Total Crime Counts by Year:")
print(yearly_counts.head())
```

```python
# playing around with 2021 data - most recent full year available
year_of_interest = '2021'
total_crime_year = yearly_counts[['WardName', 'WardCode', year_of_interest]]
total_crime_year.head()
```

```python
# merging with gdf
gdf_2021 = pd.merge(gdf, total_crime_year, left_on='NAME', right_on='WardName', how='inner').reset_index()

# checking for Barking and Dagenham
gdf_2021[gdf_2021['DISTRICT'] == 'Barking and Dagenham']
```

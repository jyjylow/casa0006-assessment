# Introduction
[[ Back to top ]](#Table-of-contents)

The study of crime and urban land use has given rise to different takes regarding the impacts of residential and commercial activity on neighbourhood crime rates. Theorists like Jane Jacobs are optimistic about the advantages of mixed-use planning, viewing them as opportunities for dispersing street activity, increasing "eyes on the street"-style social monitoring and control, and bolstering neighbourhood trust (Jacobs, 1961), all factors that can reduce the risk of crime. On the other hand, other approaches (e.g. Taylor et al., 1995) have associated mixed-use neighbourhoods with a heightened sense of anonymity, decreased participation in public space, and gaps in territoriality and street regulation (when mixing the commercial and the residential), thus increasing the propensity for crime.

Investigating the relationship between land use and crime in the context of Columbus, Ohio, Browning et al. (2010) found that it is nonlinear: while violence increased with commercial/residential density at low levels, it in fact decreased past when mixed land use increased beyond a threshold (with the exception of robbery).

### 1. Violent crime data

```python
crimeurl = "https://raw.githubusercontent.com/jyjylow/casa0006-assessment/refs/heads/main/data/MPS%20Ward%20Level%20Crime%20(Historical).csv"
response = requests.get(crimeurl)
if response.status_code == 200:
    data = StringIO(response.text)
    crimedf = pd.read_csv(data)
    print("Ward-level crime data loaded successfully.")
else:
    print(f"Failed to fetch the file: {response.status_code}")

crimedf.head()
```

```python
# excluding crimes in "Aviation Security" borough
crimedf = crimedf[crimedf.LookUp_BoroughName != "Aviation Security (SO18)"]

# filtering dataset to crimes of "Violence against the person"
crimedf = crimedf[crimedf.MajorText == "VIOLENCE AGAINST THE PERSON"]

# seeing number of wards 
#print(f"Number of wards included in this study: {len(df.WardCode.unique())}")
```

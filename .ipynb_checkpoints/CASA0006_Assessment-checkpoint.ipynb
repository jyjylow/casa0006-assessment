## Residential/commercial land use and crime
The study of crime and urban land use has given rise to different takes regarding the impacts of residential and commercial activity on neighbourhood crime rates. Theorists like Jane Jacobs are optimistic about the advantages of mixed land use, conferring the benefits of dispersing street activity, increasing "eyes on the street"-style social monitoring and control, and bolstering neighbourhood trust (Jacobs, 1961), all factors that can reduce the risk of crime. On the other hand, other approaches (e.g. Taylor, 1988) have associated the mixing of the residential and nonresidential with a heightened sense of anonymity and decreased territoriality and collective responsibility, thus increasing the propensity for crime. The latter position has contributed to general assumptions that residential land use exerts protective effects on crime, while commercial land use exacerbates it (Wo & Kim, 2020).

To what extent, however, is this assumption a simplistic take on the relationship between crime and land use? Investigating the relationship between land use and crime in the context of Columbus, Ohio, Browning et al. (2010) found it indeed nonlinear: while violent crime increased with commercial/residential density at low levels, it in fact decreased past when mixed land use increased beyond a threshold (except robbery). Similarly, Wo and Kim's (2020) study of residential versus nonresidential land use in San Francisco found a positive association between crime rates and commercial mixed-use areas, but found also that nonresidential land use exhibited non-significant and inconclusive associations with increased crime. In Zahnow's (2018) study of Brisbane, increased rates of crime was found to be tied not to mixed land use _within_ the study area, but in nearby areas, likely due to people being attracted to surrounding mixed-use areas with increased opportunities for interaction. In other words, nonresidential land use cannot always be associated with higher crime in the area (Stucky & Ottensmann, 2009), demonstrating the limitations of a binary understanding of residential/nonresidential land use and crime.

## Methodology

```python
# summing YYYYMM columns so that we have the aggregate sum of reports for past 24 months

# 1. identifying YYYYMM columns
monthly_cols = [col for col in crimedf.columns if col.isdigit()]

# 2. summing up YYYYMM columns
crime_past24_agg = pd.DataFrame()
crime_past24_agg["total"] = crimedf.groupby(['LSOA Code', 
                                             'LSOA Name'])[monthly_cols].sum().sum(axis=1)

# 3. resetting index
crime_past24_agg = crime_past24_agg.reset_index()
print("Total Violent Crime Counts (Past 24 Months):")
print(crime_past24_agg.head())
```

```python
# merging crime_past24 with LSOA shp

merged_crime24lsoa = pd.merge(gdf, crime_past24_agg, 
                           left_on='LSOA11CD', right_on='LSOA Code', 
                           how='inner')
lsoa_past24_gdf = merged_crime24lsoa[["LSOA11CD", "LSOA11NM", 
                                      "USUALRES", "geometry", "total"]]
lsoa_past24_gdf.head()
```

```python
# comparing shapes of gdfs pre- and post-merge to check if any LSOAs were not matched
print(f"No. of LSOAs in original gdf: {gdf.shape[0]}")
print(f"No. of LSOAs in merged gdf: {lsoa_past24_gdf.shape[0]}")
print(f"{gdf.shape[0] - lsoa_past24_gdf.shape[0]} LSOAs were not matched.")
```

It looks like there are some LSOAs that appear in the crime reports data but not in the LSOA shp, and vice versa, leading to less-than-perfect matching of LSOA codes between the two.

```python
# identifying which LSOAs are in crime data but not in shp 
not_in_gdf = set(crime_past24_agg['LSOA Code']) - set(gdf['LSOA11CD'])
no_match = crime_past24_agg[crime_past24_agg['LSOA Code'].isin(not_in_gdf)]
print(f"LSOAs in crime df but not in shp: {len(no_match['LSOA Code'].unique())}")

# identifying which LSOAs are in shp but not in crime data 
not_in_crimedf = set(gdf['LSOA11CD']) - set(crime_past24_agg['LSOA Code'])
no_match_crime = gdf[gdf['LSOA11CD'].isin(not_in_crimedf)]
print(f"LSOAs in shp but not in crime data: {len(no_match_crime['LSOA11CD'].unique())}")
```

```python
# Function to plot radar plots
def radar_plot_cluster_centroids(df_cluster_centroid):
    # parameters
    # df_cluster_centroid: a dataframe with rows representing a cluster centroid and columns representing variables
    
    # add an additional element to both categories and restaurants that’s identical to the first item
    # manually 'close' the line
    categories = df_cluster_centroid.columns.values.tolist()
    categories = [*categories, categories[0]]
    
    label_loc = np.linspace(start=0, stop=2 * np.pi, num=len(categories))
    
    plt.figure(figsize=(12, 8))
    plt.subplot(polar=True)
    for index, row in df_cluster_centroid.iterrows():
        centroid = row.tolist()
        centroid = [*centroid, centroid[0]]
        label = "Cluster {}".format(index)
        plt.plot(label_loc, centroid, label=label)
    lines, labels = plt.thetagrids(np.degrees(label_loc), labels=categories)
    plt.legend()
```

```python
# Visualising cluster centroids with a radar plot
df_dbscan = normed_0.copy()

# Assigning cluster numbers to LSOAs
df_dbscan['cluster_nm'] = dbsc.labels_

# Calculating variable means for each cluster as centroids 
df_dbscan_centroid = df_dbscan.groupby('cluster_nm').mean()

# Dropping outliers (DBSCAN cluster = -1)
df_dbscan_centroid.drop(-1, inplace=True)

# Plotting radar plot
radar_plot_cluster_centroids(df_dbscan_centroid)
plt.title('DBSCAN Cluster Centroid Comparison', size=20, y=1.05)
plt.show()
```

### Trying agglomerative clustering

```python
def plot_dendrogram(model, **kwargs):
    # Create linkage matrix and then plot the dendrogram

    # create the counts of samples under each node
    counts = np.zeros(model.children_.shape[0])
    n_samples = len(model.labels_)
    for i, merge in enumerate(model.children_):
        current_count = 0
        for child_idx in merge:
            if child_idx < n_samples:
                current_count += 1  # leaf node
            else:
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count

    linkage_matrix = np.column_stack(
        [model.children_, model.distances_, counts]
    ).astype(float)

    # Plot the corresponding dendrogram
    dendrogram(linkage_matrix, leaf_rotation=90., **kwargs)
```

```python
agg_cluster = AgglomerativeClustering(distance_threshold=0, 
                                      n_clusters=None).fit(normed_0)
ax = plt.gca()
plt.title("Hierarchical Clustering Dendrogram")

# plot the top three levels of the dendrogram
plot_dendrogram(agg_cluster, truncate_mode="level", p=3)
plt.xlabel("Number of points in node (or index of point if no parenthesis).")
plt.ylabel('distance')
```

```python
# Merging monthly crime data to LSOA gdf so that I can obtain rates
merged_hist = pd.merge(gdf, monthly_by_lsoa, 
                           left_on='LSOA11CD', right_on='LSOA Code', 
                           how='inner')

# Keeping relevant columns
merged_hist = merged_hist.drop(columns = ['MSOA11CD', 'MSOA11NM', 'LAD11CD', 'LAD11NM', 'RGN11CD', 'RGN11NM', 'POPDEN', 'HHOLDS', 'HHOLDRES', 'AVHHOLDSZ', 'COMESTRES', 'LSOA Name','LSOA Code']).set_index('LSOA11CD')

# Calculating monthly rate per 1000 residents
merged_hist_rate = merged_hist.copy()

for col in merged_hist_rate.columns:
    if col.isdigit():
        merged_hist_rate[col] = (merged_hist_rate[col]/merged_hist_rate['USUALRES'])*1000
merged_hist_rate.head()
```

```python
# Identifying LSOAs neighbouring Coal Drop's Yard, since we are missing crime data for Camden 022B, the actual LSOA containing Coal Drop's Yard

# Extracting row and centroid of Camden 022B LSOA
cdy_lsoa = gdf.loc[gdf['LSOA11NM'] == 'Camden 022B', 'geometry']
cdy_centroid = cdy_lsoa.geometry.centroid.values[0]

# Identifying LSOAs that have crime data AND are closest to Camden 022B
# Calculating distance between Camden 022B's centroid and centroids of LSOAs with crime data (i.e. the LSOAs that are in merged_hist_rate)
merged_hist_rate['distance'] = merged_hist_rate.geometry.centroid.distance(cdy_centroid)


# Identifying nearest LSOAs
nearest_lsoas = merged_hist_rate[['LSOA11NM', 'distance']].sort_values('distance')
print(f"3 closest LSOAs to Coal Drop's Yard:")
print(f"{nearest_lsoas.head(5)} \n")

print(f"As proxy for {poi_lsoa.iloc[0]['LSOA11NM']}, we will use {nearest_lsoas.iloc[0]['LSOA11NM']} as the treatment group in our DiD analysis of Coal Drop's Yard's impact on violent crime rates.")
```

```python
# Getting yearly average rates 

# Identifying YYYYMM columns
monthly_cols = [col for col in merged_hist.columns if col.isdigit()]

# Summing columns that start with the same year
yearly_agg = pd.DataFrame()
for year in sorted(set(col[:4] for col in monthly_cols)):
    year_cols = [col for col in monthly_cols if col.startswith(year)]
   # Calculating year totals
    yearly_agg[year] = merged_hist[year_cols].sum(axis=1)

# Adding LSOA11NM to yearly_agg
yearly_agg.insert(0, 'LSOA11NM', merged_hist['LSOA11NM'])
yearly_agg['USUALRES'] = merged_hist['USUALRES']

# Dropping 2010 and 2023 because they are incomplete years
yearly_agg = yearly_agg.drop(columns = ["2010", "2023"])

# Calculating rate from aggregates
yearly_rate = yearly_agg.copy()

for col in yearly_rate.columns:
    if col.isdigit():
        yearly_rate[col] = (yearly_rate[col]/yearly_rate['USUALRES'])*1000
yearly_rate.head()

print("Yearly crime rates:")
yearly_rate.head()
```
